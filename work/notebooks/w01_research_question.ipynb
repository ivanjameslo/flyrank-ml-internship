## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

**Research Question:** Among content pages with enough observable search activity, which pages should a content or SEO reviewer inspect first for possible refresh, protection, or monitoring?

**Decision:** The work should help determine how limited content-review time is prioritized across many pages.

**Who acts and what they do:** A content SEO reviewer would inspect the highest-priority pages and then decide whether each page needs a refresh, further investigation, protection, or monitoring.

**Cost of a wrong call:** A false priority could waste reviewer time on a page that did not need immediate attention. A missed priority could leave a genuinely declining or high-opportunity page unreviewed while its performance continues to weaken.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

visible_pages = df[df["impressions_90d"] >= 500]

print("Total pages: ", len(df))
print("Pages with at least 500 impressions in 90 days: ", len(visible_pages))
print("Share of dataset: ", f"{len(visible_pages) / len(df) * 100:.1f}%")
```

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

---

The starter dataset contains 30,000 anonymized content pages. Approximately 54.2% (16,262 pages) are currently labelled as declining in the starter snapshot. There are also 0.1% (17 pages) stale but visible pages and 43.8% (13,152 pages) declining pages with at least 100 impressions. These counts suggests that simply reviewing every potential weak page would be inefficient, so a method for prioritizing review candidates could be useful.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

declining = df["trend_direction"].str.lower().eq("down")

state_visible = (
    (df["days_since_last_update"] >= 180)
    & (df["impressions_90d"] >= 500)
)

declining_with_demand = (
    declining
    & (df["impressions_90d"] >= 100)
)

print(f"Total pages: {len(df):,}")
print(
    f"Pages currently marked as declining: "
    f"{declining.sum():,} "
    f"({declining.mean() * 100:.1f}%)"
)
print(
    f"Stale + visible pages: "
    f"{state_visible.sum():,} "
    f"({state_visible.mean() * 100:.1f}%)"
)
print(
    f"Declining pages with at least 100 impressions: "
    f"{declining_with_demand.sum():,} "
    f"({declining_with_demand.mean() * 100:.1f}%)"
)
```
